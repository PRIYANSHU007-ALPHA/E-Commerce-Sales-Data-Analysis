```python
# removing duplicate rows from the dataframe
df.drop_duplicates(inplace=True)   # removing duplicate rows from the dataframe
```

```python
df.duplicated().sum() # checking for duplicate rows in the dataframe    
```

```python
df.replace(['N/A', 'NULL', ' '], np.nan, inplace=True) # replacing 'N/A', 'NULL', and empty strings with NaN values in the dataframe
```

```python
df
```

```python
# remove extra spaces
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x) # removing leading and trailing spaces from string columns in the dataframe 
```

```python
# standardize text format 
# convert text into proper case 
# proper case means the first letter of each word is capitalized and the rest of the letters are in lowercase. For example, "hello world" would be converted to "Hello World".

df['Customer_Name'] = df['Customer_Name'].str.title() # converting the 'Customer_Name' column to proper case

df['City'] = df['City'].str.title() # converting the 'City' column to proper case

df['State'] = df['State'].str.title() # converting the 'State' column to proper case
```

```python
# validate email addresses  
# keep only the rows where the 'Email' column contains valid email addresses. A valid email address is defined as one that contains an '@' symbol and a '.' symbol after the '@' symbol.

df = df[df['Email'].str.contains('@', na=False) & df['Email'].str.contains('\.', na=False)] # for validating email addresses, we are using the str.contains() method to check if the 'Email' column contains an '@' symbol and a '.' symbol after the '@' symbol. The na=False parameter is used to exclude NaN values from the validation.

```

```python
# convert data columns
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce') # converting the 'Order_Date' column to datetime format. The errors='coerce' parameter is used to convert invalid date formats to NaT (Not a Time) values.

df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], errors='coerce') # converting the 'Delivery_Date' column to datetime format. The errors='coerce' parameter is used to convert invalid date formats to NaT (Not a Time) values.    

```

```python
# convert numerical columns
cols = ['Qty', 'Unit_Price', 'Discount'] # list of numerical columns to be converted
for c in cols:
    df[c] = pd.to_numeric(df[c], errors='coerce') # converting the columns to numeric format. The errors='coerce' parameter is used to convert invalid numerical formats to NaN values. 
```

```python
# Remove invalid quantity values
# remove records having negative or zero quantity values

df = df[df['Qty'] > 0] # keeping only the rows where the 'Qty' column has positive values
```

```python
# fill missing values 
df['Discount'].fillna(0, inplace=True) # filling missing values in the 'Discount' column with 0
df['Phone'].fillna('Unknown', inplace=True) # filling missing values in the 'Phone' column with 'Unknown'
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])  # filling missing values in the 'Delivery_Date' column with the corresponding 'Order_Date' values
```

```python
# feature engineering
# feature engineering is the process of creating new features from existing data to improve the performance of machine learning models.
# create total sales column

df['Total Sales'] = df['Qty'] * df['Unit_Price'] # creating a new column 'Sales' by multiplying the 'Qty' and 'Unit_Price' columns to calculate the total sales for each order
df.head() # displaying the first 5 rows of the dataframe to verify the new 'Sales' column has been added correctly
```

```python
# net amount column
df['Net Amount'] = df['Total Sales'] -(df['Total Sales'] * df['Discount']/100) # creating a new column 'Net Amount' by subtracting the discount amount from the 'Total Sales' column. The discount amount is calculated by multiplying the 'Total Sales' column with the 'Discount' column divided by 100 to convert the percentage to a decimal.
df.head() # displaying the first 5 rows of the dataframe to verify the new 'Net Amount' column has been added correctly

```

```python
# profit column 20%
df['Profit'] = df['Net Amount'] * 0.2 # creating a new column 'Profit' by multiplying the 'Net Amount' column with 0.2 to calculate the profit for each order   
df.head() # displaying the first 5 rows of the dataframe to verify the new 'Profit' column has been added correctly
```

```python
# order month column
df['Order_Month'] = df['Order_Date'].dt.month # creating a new column 'Order_Month' by extracting the month from the 'Order_Date' column using the dt accessor and the month attribute
```

```python
# order year column
df['Order_Year'] = df['Order_Date'].dt.year # creating a new column 'Order_Year' by extracting the year from the 'Order_Date' column using the dt accessor and the year attribute
```

```python
# order weekday column
df['weekday'] = df['Order_Date'].dt.day_name() # creating a new column 'Order_Weekday' by extracting the weekday name from the 'Order_Date' column using the dt accessor and the day_name() method
df.head() # displaying the first 5 rows of the dataframe to verify the new 'Order_Weekday' column has been added correctly
```

```python
# final data validation 
df.info() # displaying information about the dataframe, including column names, data types, and non-null value counts
```

```python
df.isnull().sum() # checking for missing values in each column of the dataframe after data cleaning and feature engineering
```

```python
from sqlalchemy import create_engine

engine = create_engine("postgresql+psycopg2://postgres:1234@localhost:5432/ecommerce") # creating a connection to the PostgreSQL database using SQLAlchemy. The connection string specifies the database type (PostgreSQL), the driver (psycopg2), the username (postgres), the password (1234), the host (localhost), and the database name (ecommerce).

df.to_sql("ecommerce_data",engine,if_exists="replace",index=False) # writing the DataFrame to a PostgreSQL table named 'ecommerce_data'

print("Data imported successfully!") # printing a message to indicate that the data has been imported successfully into the PostgreSQL database
```
